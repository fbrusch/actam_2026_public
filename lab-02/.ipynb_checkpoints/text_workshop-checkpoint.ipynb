### Text that contains HTML characters

Suppose the heading itself should say `<hello>`. Putting that text directly inside our tags can make the renderer treat it as markup.

Python supplies `escape` to represent special characters as text. Look at both the string and the rendered output:

````python
from html import escape

print(escape("<hello> & goodbye"))
display(HTML("<h2>" + escape("<hello> & goodbye") + "</h2>"))
````

### Keeping only some elements

A comprehension can include a condition after its loop. Only elements for which the condition is true contribute to the result.

This example keeps positive numbers:

````python
numbers = [-2, 0, 3, 7]
print([number for number in numbers if number > 0])
````
