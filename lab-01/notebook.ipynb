## A broad hill
<img class="exercise-plot" src="./plots/hill.svg" alt="Twenty-one discrete samples forming a smooth, symmetric hill from zero up to one and back to zero" />

<p class="exercise-prompt">Can one expression create both sides of the hill?</p>

The curve is a **downward parabola** sampled at 21 integer points,
from `x = 0` to `x = 20`.

### The equation

$$y = -\frac{x^2}{100} + \frac{x}{5}$$

that is $a = -1/100$, $b = 1/5$, $c = 0$ in the canonical form
$y = ax^2 + bx + c$.

Each coefficient controls one thing:

- **$a < 0$** makes the parabola open downward. Its magnitude sets how fast the curve falls away:
  the smaller $|a|$, the broader the hill.
- **$b$** shifts the peak sideways. The vertex sits at
  $x = -\dfrac{b}{2a} = -\dfrac{1/5}{-1/50} = 10$, the centre of
  the sampled range.
- **$c = 0$** is the value at $x = 0$, so the curve starts at the
  origin.

### Two ways to build the list

**Hard-coded values**: always works, including for experimental
data with no rule behind it, but the numbers are opaque: the
symmetry is invisible, and changing the resolution means
recomputing everything.

**List comprehension**: a single expression generates both sides.
Moving to a different resolution takes changing the coefficients.

````python
# Hard-coded values: the less convenient way

values_ex_1_hardcoded = [
    0.00, 0.19, 0.36, 0.51, 0.64, 0.75, 0.84, 0.91, 0.96, 0.99, 1.00, 
    0.99, 0.96, 0.91, 0.84, 0.75, 0.64, 0.51, 0.36, 0.19, 0.00
]
plt.figure(figsize=(9, 2.5))
plt.plot(values_ex_1_hardcoded, marker='o')
plt.xlabel('Sample index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
````

````python
# List comprehension: the more convenient way
values_ex_1 = [-x**2/100 + x/5 for x in range(21)]
plt.figure(figsize=(9, 2.5))
plt.plot(values_ex_1, marker='o')
plt.xlabel('Sample index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
````

# A narrow peak

<img class="exercise-plot" src="./plots/spike.svg" alt="Twenty-one discrete samples with one narrow peak near the middle" />

<p class="exercise-prompt">What changes quickly near the center and slowly far away?</p>

````python
# Hard-coded values

values_ex_2_hard = [
    0.01, 0.01, 0.02, 0.02, 0.03, 0.04, 0.06, 0.10, 0.20, 0.50,
    1.00, 0.50, 0.20, 0.10, 0.06, 0.04, 0.03, 0.02, 0.02, 0.01, 0.01
]
plt.figure(figsize=(9, 2.5))
plt.plot(values_ex_2_hard, marker='o')
plt.xlabel('Sample index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
````

````python
# List comprehension

values_ex_2 = [1 / (1 + (x - 10)**2) for x in range(21)]
plt.figure(figsize=(9, 2.5))
plt.plot(values_ex_2, marker='o')
plt.xlabel('Sample index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
````

# Two unequal peaks

<img class="exercise-plot" src="./plots/two-peaks.svg" alt="Thirty discrete samples with a small peak near index seven and a taller peak near index twenty" />

<p class="exercise-prompt">Can two simple shapes add up to this one?</p>

````python
values_ex_3 = [1/(1 + (x - 7)**2) + 2/(1 + (x - 20)**2) for x in range(30)]
plt.figure(figsize=(9, 2.5))
plt.plot(values_ex_3, marker='o')
plt.xlabel('Sample index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
````

````python
# Parametrized version
peaks = [(7, 1), (20, 2)]
values_ex_3_parametrized = [sum(A/(1 + (x - c)**2) for c, A in peaks) for x in range(30)]
plt.figure(figsize=(9, 2.5))
plt.plot(values_ex_3, marker='o')
plt.xlabel('Sample index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
````
