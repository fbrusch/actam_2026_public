### E5 · What does the format permit?
*(compare: lab-02 E7)*

**Explore · Inspect assumptions**

Same rule as E3: exactly one colon, a nonempty name, a nonnegative integer.

For each input, decide whether it follows the rule and predict what the basic `readRecord` does:

```text
Ada: 12
Ada:
: 12
Ada: 12: extra
Ada: -2
Ada 12
```

Write `checkedRecord(line)` with checks for the number of fields, an empty name, and a negative number. Use `if` and `throw new Error("a helpful message")`. Let `Number`/`parseInt` produce `NaN` for invalid integer text, and check for that too.

````typescript
function checkedRecord(line) {
    let parts = line.split(":")

    if (parts.length !== 2) {
        throw new Error(`expected exactly one colon, got: "${line}"`)
    }

    let name = parts[0].trim()
    let numberText = parts[1].trim()

    if (name === "") {
        throw new Error(`name cannot be empty: "${line}"`)
    }

    let pages = Number(numberText)

    if (Number.isNaN(pages)) {
        throw new Error(`expected a number, got: "${numberText}"`)
    }
    if (pages < 0) {
        throw new Error(`expected a nonnegative number, got: ${pages}`)
    }

    return [name, pages]
}

console.log(checkedRecord("Ada: 12"))

let badInputs = ["Ada:", ": 12", "Ada: 12: extra", "Ada: -2", "Ada 12"]

for (let bad of badInputs) {
    try {
        checkedRecord(bad)
    } catch (e) {
        console.log(`"${bad}" ->`, e.message)
    }
}
````
