## Patterns

One track per instrument. Replace this example with what you hear.

```python
# id: another_one_bites
bpm = 110
step_s = 60 / bpm / 4
#          1 e & a 2 e & a 3 e & a 4 e & a
tracks = [('x...x...x...x...',kick,step_s),
          ('....x.......x...',snare,step_s),
          ('x.x.x.x.x.x.x.x.',closed_hat,step_s),
          ('................',open_hat,step_s)]

another_one_bites = render_tracks(tracks)
```

```python
# id: be_my_baby
bpm = 114
step_s = 60 / bpm / 4
tracks = [('............x...',kick,step_s),
          ('x.....x.x...x...',snare,step_s),
          ('............x...',closed_hat,step_s),
          ('............x...',open_hat,step_s)]

be_my_baby = render_tracks(tracks)
```

```python
# id: superstition
bpm = 100
step_s = 60 / bpm / 4
#           1 e & a 2 e & a 3 e & a 4 e & a
tracks = [('........x...x...x...x...x...x...x...x...x...x...x...x...x...x...x...x...',kick,step_s),
          ('x...x.xx....x.......x.......x.......x.......x.......x.......x.......x...',snare,step_s),
          ('........x.x.x.xxxxx.x.xxx.x.x.xxx.xxxxx.x.xxx.xxx.xxxxxxx.x.x.xxx.x.xxxx',closed_hat,step_s),
          ('............x.......x.......x.......x.......x.......x.......x.......x...',open_hat,step_s)]

superstition = render_tracks(tracks)
```

## Listen

```python
play(another_one_bites * 4)
```
